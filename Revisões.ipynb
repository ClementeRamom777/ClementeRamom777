```python
jogo = 'Alan Wake 2'
preco = 349.00
percentual = 10

desconto = preco * percentual/100
final = preco - desconto
print('%s custa R$ %.2f com desconto de R$ %.2f.' % (jogo, final, desconto))
print('{} custa R$ {:.2f} com desconto de R$ {:.2f}.'.format(jogo, final, desconto))
print(f'{jogo} custa R$ {final:.2f} com desconto de R$ {desconto:.2f}.')
```

```python
game = input('Digite o nome do jogo: ')
valor = float(input('Digite o preço do jogo: '))
percentual_des = float(input('Digite o percentual de desconto: '))
parcelas = int(input('Digite o número de parcelas: '))

desconto_jogo = valor * percentual_des/100
valor_final = valor - desconto_jogo
valor_parcela = valor_final / parcelas
limite_parcela = valor_parcela <= 100

print('%s: de R$ %.2f por R$ %.2f.' % (game, valor, valor_final))
print('%dx de R$ %.2f.' % (parcelas, valor_parcela))
print('Parcela cabe no limite de R$ 100,00? %s.' % (limite_parcela))
```

```python
total_conta = float(input('Digite o valor total da conta: '))
clientes = int(input('Digite o número de clientes: '))

conta_individual = total_conta / clientes
print('Entrada: %.2f e %d'% (total_conta, clientes))
print('Cada um paga R$ %.2f' % (conta_individual))
if conta_individual > 50:
  print('Acima do combinado')
```

```python
nota_1 = float(input('Digite a primeira nota: '))
nota_2 = float(input('Digite a segunda nota: '))

print('Entrada: %.1f e %.1f' % (nota_1, nota_2))
media_final = (nota_1 + nota_2) / 2
print('Media: %.1f' % (media_final))
if media_final >= 7:
  print('Aprovado!')
else:
  print('Reprovado!')
```

```python
idade = int(input('Digite sua idade: '))
renda = float(input('Digite sua renda: '))
anos_emprego = int(input('Digite quantos anos você está empregado: '))

print('Entrada: %d, %.2f e %d' % (idade, renda, anos_emprego))
if idade >= 18 and renda >= 2000 and anos_emprego >= 1:
  print('Empréstimo aprovado!')
else:
  print('Empréstimo negado!')
```
